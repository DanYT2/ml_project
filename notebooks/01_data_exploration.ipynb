## 1. Dataset Overview

First, let's check if the dataset exists and explore its structure.


```python
# Dataset path
DATA_DIR = Path('../data')
TRAIN_DIR = DATA_DIR / 'asl_alphabet_train' / 'asl_alphabet_train'

# Check if data exists
if not TRAIN_DIR.exists():
    print(f"Dataset not found at {TRAIN_DIR}")
    print("Please run: python scripts/download_dataset.py")
else:
    print(f"Dataset found at {TRAIN_DIR}")
    
    # List classes
    classes = sorted([d.name for d in TRAIN_DIR.iterdir() if d.is_dir()])
    print(f"\nClasses ({len(classes)}): {classes}")
    
    # Count images per class
    class_counts = {}
    for class_name in classes:
        class_dir = TRAIN_DIR / class_name
        count = len(list(class_dir.glob('*.jpg')))
        class_counts[class_name] = count
    
    print(f"\nTotal images: {sum(class_counts.values()):,}")
```
